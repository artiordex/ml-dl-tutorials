```python
# PyTorch 메인 패키지 (텐서 연산, 자동 미분 등 핵심 기능 제공)
import torch

# 신경망 모델을 구성하기 위한 모듈 (Linear, Conv 층 등)
import torch.nn as nn

# 활성화 함수(ReLU, Softmax 등)와 신경망 관련 함수 모음
import torch.nn.functional as F

# 최적화 알고리즘(Adam, SGD 등) 제공
import torch.optim as optim

# Dataset, DataLoader 등 데이터 처리 도구 제공
import torch.utils.data as data_utils
```

```python
# 손글씨 숫자(Digits) 분류를 위한 간단한 신경망 모델 정의
class DigitsModel(nn.Module):
    def __init__(self, input_size, output_size):
        super(DigitsModel, self).__init__()

        # 입력층 → 은닉층(128 노드) → 출력층 구조의 MLP 모델
        self.model = nn.Sequential(
            nn.Linear(input_size, 128),  # 입력 → 은닉층
            nn.ReLU(),                   # 활성화 함수
            nn.Linear(128, output_size)  # 은닉층 → 출력층 (0~9 클래스)
        )
        
    def forward(self, x):
        # 순전파 계산
        output = self.model(x)

        # 예측 클래스 계산 (gradient 불필요)
        with torch.no_grad():
            value = torch.argmax(output, dim=1)

        # output: 로짓 값, value: 예측된 클래스(0~9)
        return output, value
```

```python
X = torch.Tensor(data.data)
y = torch.LongTensor(data.target)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) # train과 test를 8:2로 분할
```

```python
model = DigitsModel(64, 10)
```

```python
# 추론 모드 (gradient 계산 비활성화)
with torch.no_grad():

    # X_train[0]은 단일 샘플이므로 배치 차원(1)을 추가한 후 모델에 입력
    # → unsqueeze(dim=0): (features) → (1, features)
    y, _ = model(torch.unsqueeze(X_train[0], dim=0))

    # 모델 출력(logits) 확인
    print(y)

    # 출력값의 총합 (단순 디버깅용)
    print(y.sum())

    # 모델 예측 클래스(np.argmax)와 실제 정답(y_train[0]) 비교
    print(np.argmax(y), y_train[0])
```

```python
# 모델 파라미터를 Adam 최적화 알고리즘으로 업데이트하도록 설정
# lr=0.001 → 학습률(learning rate)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
# 입력 64차원(8×8 이미지 펼침), 출력 10개(0~9 숫자)로 모델 생성
model = DigitsModel(64, 10)

# 모델 파라미터를 Adam optimizer로 업데이트 (학습률 0.001)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 다중 클래스 분류에서 사용하는 손실 함수
criterion = nn.CrossEntropyLoss()

# 학습 반복 횟수 설정
n_epochs = 10000

avg_loss = 0

for epoch in range(n_epochs):

    # 모델 예측값 계산
    y, _ = model(X_train)

    # digits 데이터셋의 레이블은 이미 0~9이므로 그대로 사용
    target = y_train

    # 손실값 계산 (배치 전체의 손실을 합산)
    loss = criterion(y, target).sum()

    # 기존 gradient 초기화
    optimizer.zero_grad()

    # 역전파 수행하여 gradient 계산
    loss.backward()

    # 계산된 gradient로 모델 파라미터 업데이트
    optimizer.step()

    # 1000 epoch마다 학습 손실 출력
    if epoch % 1000 == 0:
        print('epoch {}th loss: {}'.format(epoch, loss.data))
```

```python
# 입력 64차원(8×8 이미지), 출력 10개(0~9 숫자)로 모델 생성
model = DigitsModel(64, 10)

# Adam optimizer 설정 (학습률 0.004)
optimizer = torch.optim.Adam(model.parameters(), lr=0.004)

# 다중 클래스 분류용 손실 함수
criterion = nn.CrossEntropyLoss()

# 학습 반복 횟수
n_epochs = 2000

# 손실 기록용 리스트
list_training_loss = []
list_test_loss = []

for epoch in range(n_epochs):

    # 학습 단계
    model.train()                 # 학습 모드
    y, _ = model(X_train)         # 학습 데이터 예측
    target = y_train              # digits 레이블은 0~9이므로 그대로 사용
    loss = criterion(y, target).sum()  # 학습 손실 계산

    optimizer.zero_grad()         # 기존 gradient 초기화
    loss.backward()               # 역전파 계산
    optimizer.step()              # 파라미터 업데이트

    # 평가 단계 (200 epoch마다)
    if epoch % 200 == 0:
        model.eval()              # 평가 모드
        y_pred, _ = model(X_test) # 테스트 데이터 예측
        test_loss = criterion(y_pred, y_test).sum()  # 테스트 손실

        # 손실 값 출력
        print('epoch {}th training loss: {} test loss: {}'.format(
            epoch, loss.data, test_loss.data))

        # 손실 기록 저장
        list_training_loss.append(loss.data)
        list_test_loss.append(test_loss.data)
```

```python
import matplotlib.pyplot as plt

plt.plot(list_training_loss, label='training')
plt.plot(list_test_loss, label='test')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import accuracy_score

# 추론 모드: gradient 계산 비활성화
with torch.no_grad():

    # 모델 예측 수행 (y_pred는 예측된 클래스 인덱스)
    _, y_pred = model(X_test)

    # accuracy_score: 실제 레이블(y_test)과 예측값(y_pred)을 비교해 정확도 계산
    score = accuracy_score(y_test, y_pred)

    # 정확도 출력
    print(score)
```
