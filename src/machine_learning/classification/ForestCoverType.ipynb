```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torch.utils.data as data_utils
```

```python
# Cover Type 분류를 위한 간단한 신경망 모델 정의
class CovTypeModel(nn.Module):
    def __init__(self, input_size, output_size):
        super(CovTypeModel, self).__init__()

        # 입력층 → 은닉층(128) → 출력층 구조의 기본 MLP
        self.model = nn.Sequential(
            nn.Linear(input_size, 128),  # 입력 → 은닉층
            nn.ReLU(),                   # 활성화 함수
            nn.Linear(128, output_size)  # 은닉층 → 출력층 (클래스 개수만큼)
        )
        
    def forward(self, x):
        # 순전파 계산
        output = self.model(x)

        # 예측 클래스(argmax)를 계산 (학습시 gradient 불필요)
        with torch.no_grad():
            value = torch.argmax(output, dim=1)

        # output: 로짓 값, value: 예측된 클래스 인덱스
        return output, value
```

```python
# scikit-learn의 feature 데이터를 PyTorch Tensor(float32)로 변환
# → 신경망 입력으로 사용하기 위한 형태
X = torch.Tensor(data.data)

# target(label)을 PyTorch LongTensor(int64)로 변환
# → CrossEntropyLoss 등 분류 손실함수에서 요구하는 정수 클래스 형태
y = torch.LongTensor(data.target)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

# 데이터를 학습용(80%)과 테스트용(20%)으로 분할
# → 모델 평가를 위해 train/test 데이터를 분리하는 필수 단계
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# 입력 특징 54개, 출력 클래스 7개를 갖는 모델 생성
model = CovTypeModel(54, 7)
```

```python
# 학습 모드가 아닌 추론(inference) 모드에서 실행 (gradient 계산 비활성화)
with torch.no_grad():

    # X_train[0]은 1차원 벡터이므로 배치 차원을 추가해 모델 입력 형태로 변환
    # → unsqueeze(dim=0): (features) → (1, features)
    y, _ = model(torch.unsqueeze(X_train[0], dim=0))

    # 모델의 raw output(logits) 출력
    print(y)

    # 출력 벡터의 합 (단순 확인용)
    print(y.sum())

    # np.argmax(y): 모델이 예측한 클래스
    # y_train[0] - 1: 실제 정답(데이터셋 레이블이 1~7일 경우 보정)
    print(np.argmax(y), y_train[0] - 1)
```

```python
# 모델 생성 (입력 54, 출력 클래스 7)
model = CovTypeModel(54, 7)

# Adam optimizer 사용, 학습률 0.001
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 다중 클래스 분류에 사용하는 손실 함수
criterion = nn.CrossEntropyLoss()

# 전체 학습 반복 횟수(epochs)
n_epochs = 1000

avg_loss = 0

for epoch in range(n_epochs):

    # 모델 예측값(logits) 계산
    y, _ = model(X_train)

    # 레이블이 1~7일 경우 CrossEntropyLoss는 0~6을 요구하므로 -1 보정
    target = y_train - 1

    # 배치 전체의 손실을 계산한 후 합산
    loss = criterion(y, target).sum()

    # 이전 gradient 초기화
    optimizer.zero_grad()

    # loss 기준으로 gradient 계산
    loss.backward()

    # optimizer가 파라미터 업데이트 수행
    optimizer.step()

    # 진행 상황 출력 (100 epoch마다)
    if epoch % 100 == 0:
        print('epoch {}th loss: {}'.format(epoch, loss.data))
```

```python
# 확인하고 싶은 학습 데이터의 인덱스
n = 99  

# 추론 모드 (gradient 계산 비활성화)
with torch.no_grad():

    # X_train[n]은 단일 샘플이므로 배치 차원을 추가해 (1, feature) 형태로 변환
    y, _ = model(torch.unsqueeze(X_train[n], dim=0))

    # 모델의 출력(logits) 확인
    print(y)

    # 출력 벡터의 합 (debugging 용)
    print(y.sum())

    # np.argmax(y): 모델이 예측한 클래스
    # y_train[n] - 1: 실제 정답(label)을 CrossEntropyLoss 형태(0~6)와 동일하게 보정한 값
    print(np.argmax(y), y_train[n] - 1)
```

```python
# 모델 생성 (입력 54개, 클래스 7개)
model = CovTypeModel(54, 7)

# Adam 최적화 알고리즘 설정 (lr=0.004)
optimizer = torch.optim.Adam(model.parameters(), lr=0.004)

# 다중 클래스 분류에 사용하는 손실 함수
criterion = nn.CrossEntropyLoss()

# 학습 반복 횟수
n_epochs = 1000

# 학습/테스트 손실 기록용 리스트
list_training_loss = []
list_test_loss = []

for epoch in range(n_epochs):

    # 학습 단계
    model.train()  # 학습 모드
    y, _ = model(X_train)       # 예측값 계산
    target = y_train - 1        # 타깃을 0~6 범위로 맞춤
    loss = criterion(y, target).sum()  # 학습 손실 계산

    optimizer.zero_grad()       # 기존 gradient 초기화
    loss.backward()             # 역전파 수행
    optimizer.step()            # 모델 파라미터 업데이트

    # 평가 단계 (100 epoch마다)
    if epoch % 100 == 0:
        model.eval()  # 평가 모드
        y_pred, _ = model(X_test)
        test_loss = criterion(y_pred, y_test - 1).sum()

        # 진행 상황 출력
        print('epoch {}th training loss: {} test loss: {}'.format(
            epoch, loss.data, test_loss.data))

        # 손실 기록
        list_training_loss.append(loss.data)
        list_test_loss.append(test_loss.data)
```

```python
import matplotlib.pyplot as plt

# 기록해둔 학습 손실(training loss)을 선 그래프로 시각화
plt.plot(list_training_loss, label='training')

# 기록해둔 테스트 손실(test loss)을 선 그래프로 시각화
plt.plot(list_test_loss, label='test')

# 그래프 범례 표시
plt.legend()

# 그래프 출력
plt.show()
```

```python
from sklearn.metrics import accuracy_score

# 추론 모드에서 테스트 데이터 예측 수행 (gradient 비활성화)
with torch.no_grad():

    # model(X_test)는 (logits, predicted_class) 두 값을 반환
    _, y_pred = model(X_test)

    # accuracy_score: 실제값과 예측값을 비교해 정확도 계산
    # y_test는 1~7 범위이므로, 모델 학습과 동일하게 0~6 범위로 맞추기 위해 -1 적용
    score = accuracy_score(y_test - 1, y_pred)

    # 정확도 출력
    print(score)
```
