# [Gradient Boosting - classification](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.GradientBoostingClassifier.html#sklearn.ensemble.GradientBoostingClassifier)

```python
from sklearn.svm import SVC
import sklearn.ensemble
from sklearn import datasets
import pandas as pd
import seaborn as sns
import numpy as np
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
data = datasets.load_iris()
```

```python
X_data = data.data
y_data = data.target
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2) # train과 test를 8:2로 분할
```

```python
model = sklearn.ensemble.GradientBoostingClassifier(random_state=0)
```

```python
model.fit(X_train, y_train)
```

```python
y = model.predict(X_train)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import axes3d, Axes3D 

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X_train[:,0], X_train[:,1], X_train[:,2], c = y_train)
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import axes3d, Axes3D 

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X_train[:,0], X_train[:,1], X_train[:,2], c = y)
plt.show()
```

```python
print(model.score(X_train, y_train))
```

```python
train_score = []
test_score = []
n_range = range(1, 100)

for n in n_range:
    model = sklearn.ensemble.GradientBoostingClassifier(n_estimators=n, random_state=0)
    model.fit(X_train, y_train)
    train_score.append(model.score(X_train, y_train))
    test_score.append(model.score(X_test, y_test))

plt.figure()
plt.plot(n_range, train_score, label='train')
plt.plot(n_range, test_score, label='test')
plt.legend()
plt.xlabel('n_estimators')
plt.ylabel('score')
plt.show()
```
