```python
import sklearn.datasets
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
# sklearn에서 예제 이미지 두 장을 포함한 샘플 이미지 데이터셋을 불러온다.
data = sklearn.datasets.load_sample_images()

# 데이터셋이 가진 키들을 확인한다.
print(data.keys())

# 각 키에 대해 반복하면서 타입과 shape을 출력한다.
for f in data.keys():
    t = type(data[f])
    print('key: {}, type: {}'.format(f, t))

    # 만약 값이 numpy 배열(ndarray)이면 이미지 데이터이므로 shape를 출력한다.
    if t == np.ndarray:
        print('shape: {}'.format(data[f].shape))
```
